```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
pf=pd.read_csv('Algerian.csv')
```

```python
pf.head()
```

```python
pf.columns
```

```python
### drop month,day and year
pf.drop(['month','day','year'],axis=1,inplace=True)
```

```python
pf.head()
```

```python
pf['Classes'].value_counts()
```

```python
### encoding
pf['Classes']=np.where(pf['Classes'].str.contains('not fire'),0,1)
## pf['Classes']=np.where(pf['Classes'].str.contains('distracted'),1,pf['Classes'])
```

```python
pf.tail()
```

```python
pf['Classes'].value_counts()
```

```python
### independent and dependent feature
X=pf.drop('FWI',axis=1)
y=pf['FWI']
```

```python
X.head()
```

```python
y.head()
```

```python
### train test split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
### feature selection based on correlation
X_train.drop('Region',axis=1).corr()
```

```python
### multicolllinarity
plt.figure(figsize=(12,10))
corr=X_train.drop('Region',axis=1).corr()
sns.heatmap(X_train.drop('Region',axis=1).corr(),annot=True)
```

```python
def correlation(dataset,threshold):
    col_corr=set()
    corr_matrix=dataset.corr()
    for i in range(len(corr_matrix.columns)):
        for j in range(i):
            if abs(corr_matrix.iloc[i,j])>threshold:
                colname=corr_matrix.columns[i]
                col_corr.add(colname)
    return col_corr
```

```python
### threshold --Domain experteries
correlation(X_train.drop('Region',axis=1),0.85)
```

```python
### drop this features when corelation is more than 0.85
corr_features = correlation(X_train.drop('Region',axis=1),0.85)
X_train.drop(corr_features,axis=1,inplace=True)
X_test.drop(corr_features,axis=1,inplace=True)
X_train.shape,X_test.shape
```

```python
### feature scaling or standarlization
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()

X_train_processed = X_train.drop('Region', axis=1)
X_test_processed = X_test.drop('Region', axis=1)

X_train_scale=sc.fit_transform(X_train_processed)
X_test_scale=sc.transform(X_test_processed)
```

```python
X_train_scale
```

```python
### box plot to understand effect of standard scalar
plt.subplots(figsize=(12,8))
plt.subplot(1,2,1)
sns.boxplot(X_train_scale)
plt.title('Boxplot before standard scalar')
plt.subplot(1,2,2)
sns.boxplot(data=X_train_scale)
plt.title('Boxplot after standard scalar')
plt.show()
```

```python
### linear model regression
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
lr=LinearRegression()
lr.fit(X_train_scale,y_train)
y_pred=lr.predict(X_test_scale)
mse=mean_squared_error(y_test,y_pred)
mae=mean_absolute_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)
print('MSE',mse)
print('MAE',mae)
print('R2',r2)
```

```python
plt.scatter(y_test,y_pred)
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.show()
```

```python
### Lasso regression model
from sklearn.linear_model import Lasso
lasso=Lasso()
lasso.fit(X_train_scale,y_train)
y_pred=lasso.predict(X_test_scale)
mse=mean_squared_error(y_test,y_pred)
mae=mean_absolute_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)
print('MSE',mse)
print('MAE',mae)
print('R2',r2)
```

```python
plt.scatter(y_test,y_pred)
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.show()
```

```python
### Ridge regression
from sklearn.linear_model import Ridge
ridge=Ridge()
ridge.fit(X_train_scale,y_train)
y_pred=ridge.predict(X_test_scale)
mse=mean_squared_error(y_test,y_pred)
mae=mean_absolute_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)
print('MSE',mse)
print('MAE',mae)
print('R2',r2)
```

```python
### elastic net
from sklearn.linear_model import ElasticNet
elastic=ElasticNet()
elastic.fit(X_train_scale,y_train)
y_pred=elastic.predict(X_test_scale)
mse=mean_squared_error(y_test,y_pred)
mae=mean_absolute_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)
print('MSE',mse)
print('MAE',mae)
print('R2',r2)
```

```python
plt.scatter(y_test,y_pred)
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.show()
```
